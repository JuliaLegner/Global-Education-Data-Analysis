```python

# Question 1f

# year columns
yearcolumn = [
    'year_2010', 'year_2011', 'year_2012', 'year_2013', 'year_2014',
    'year_2015', 'year_2016', 'year_2017', 'year_2018', 'year_2019',
    'year_2020', 'year_2021', 'year_2022', 'year_2023', 'year_2024'
]

# transform year column
trans_rows = pd.melt(df_education_clean, id_vars=['Country', 'ISO_code', 'Education_Indicator', 'Indicator_Code'], value_vars=yearcolumn, var_name='year', value_name='value')

# Clean years
trans_rows['year'] = trans_rows['year'].str.split('_').str[1].astype(int)

# categorising data
def education_level(indicate):
    if pd.isna(indicate):
        return None
    text = indicate.lower()
    if "pre-primary" in text:
        return "pre-primary"
    elif "secondary" in text:
        return "secondary"
    elif "primary" in text:
        return "primary"
    else:
        return None


def metric_type(indicate):
    if pd.isna(indicate):
        return None
    text = indicate.lower()
    if "teachers" in text:
        return "teachers"
    elif "enrolment" in text:
        return "enrolment"
    else:
        return None

#categorising functions
trans_rows['education_level'] = trans_rows['Education_Indicator'].apply(education_level)
trans_rows['metric_type'] = trans_rows['Education_Indicator'].apply(metric_type)

#remove nan/null
trans_rows = trans_rows[trans_rows['education_level'].notnull()]
trans_rows = trans_rows[trans_rows['metric_type'].notnull()]

#values=numeric
trans_rows['value'] = pd.to_numeric(trans_rows['value'], errors='coerce')

# split into teachers and enrolment tables
teachers = (trans_rows[trans_rows['metric_type'] == 'teachers']
    .copy()
    .rename(columns={'value': 'teachers'})
)

enrolment = (
    trans_rows[trans_rows['metric_type'] == 'enrolment']
    .copy()
    .rename(columns={'value': 'enrolment'})
)

#merge
ratio_df = teachers.merge(
    enrolment,
    on=['Country', 'ISO_code', 'year', 'education_level'],
    how='inner'
)

# calculate ratios
ratio_df['ratio_teacherstudent'] = ratio_df['teachers'] / ratio_df['enrolment']
ratio_df['ratio_studentteacher'] = ratio_df['enrolment'] / ratio_df['teachers']

# display results
countries_to_show = ['Switzerland', 'Germany']

ratio_two_countries = ratio_df[ratio_df['Country'].isin(countries_to_show)].copy()
ratio_two_countries = ratio_two_countries.sort_values(
    by=['Country', 'education_level', 'year']
)

#output
nice_output = ratio_two_countries[[
    'Country', 'year', 'education_level',
    'teachers', 'enrolment',
    'ratio_teacherstudent', 'ratio_studentteacher'
]].round({'ratio_teacherstudent': 4, 'ratio_studentteacher': 4})

print(len(nice_output))


```

```python
#question 1l)

# Start from df_country
country_df = df_country.copy()

#numeric values
country_df["population"] = country_df["population"].astype(float)
country_df["area"] = country_df["area"].astype(float)

#nan, missing zero values
country_df = country_df.dropna(subset=["population", "area"])
country_df = country_df[country_df["area"] > 0]

#population density
country_df["density"] = country_df["population"] / country_df["area"]

#density class
country_df["density_class"] = "sparse"

#density 100–500 people
country_df.loc[
    (country_df["density"] >= 100) & (country_df["density"] <= 500),
    "density_class"
] = "dense"

#density >500
country_df.loc[
    country_df["density"] > 500,
    "density_class"
] = "very_dense"

#sorting
country_df = country_df.sort_values(
    by=["region", "density"],
    ascending=[True, False]
)

#top 5 countries
top5_by_region = (
    country_df
    .groupby("region", group_keys=False)
    .head(5)
)

#output
top5_by_region = top5_by_region[
    ["region", "country", "population", "area", "density", "density_class"]
]

top5_by_region.head()
```

```python
#question 1m

#literacy 2019
literacy_rate1 = df_education_clean[
    df_education_clean['Education_Indicator'] ==
    'Adult literacy rate, population 15+ years, both sexes (%)'
][['Country', 'ISO_code', 'year_2019']].copy()

literacy_rate1 = literacy_rate1.rename(columns={'year_2019': 'literacy_rate'})

#population
population = df_country[['country', 'population']].copy()
population = population.rename(columns={'country': 'Country'})

#merge
data1 = literacy_rate1.merge(population, on='Country', how='inner')

#numeric
data1['literacy_rate'] = data1['literacy_rate'].astype(float)
data1['population'] = data1['population'].astype(float)

#65-80%
data1['adult_pop_min'] = data1['population'] * 0.65
data1['adult_pop_max'] = data1['population'] * 0.80

data1['illiteracy_rate'] = 100 - data1['literacy_rate']
data1['illiterates_min'] = data1['adult_pop_min'] * data1['illiteracy_rate'] / 100
data1['illiterates_max'] = data1['adult_pop_max'] * data1['illiteracy_rate'] / 100

#clean
outcome = data1.dropna(subset=['illiterates_min', 'illiterates_max']).copy()

outcome['population'] = outcome['population'].round(0)
outcome['illiterates_min'] = outcome['illiterates_min'].round(0)
outcome['illiterates_max'] = outcome['illiterates_max'].round(0)
outcome['literacy_rate'] = outcome['literacy_rate'].round(2)
outcome['illiteracy_rate'] = outcome['illiteracy_rate'].round(2)

#output
outcome = outcome.sort_values('illiterates_max', ascending=False)

outcome_display = outcome[[
    'Country',
    'population',
    'literacy_rate',
    'illiteracy_rate',
    'illiterates_min',
    'illiterates_max'
]]

outcome_display.head()
```

```python
# Question 1n

#secondary enrolment 2015
secondary_enrolment = df_education_clean[df_education_clean['Education_Indicator'] =='Enrolment in secondary education, both sexes (number)'][['Country', 'year_2015']].copy()
secondary_enrolment = secondary_enrolment.rename(columns={'year_2015': 'enrolment_2015'})

#numeric
secondary_enrolment['enrolment_2015'] = secondary_enrolment['enrolment_2015'].astype(float)

#regions df_country
regions = df_country[['country', 'region']].copy()
regions = regions.rename(columns={'country': 'Country', 'region': 'Region'})

#merge
data = secondary_enrolment.merge(regions, on='Country', how='inner')

#missing
data = data.dropna(subset=['enrolment_2015'])

#average/mean
avg_enrol = data.groupby('Region')['enrolment_2015'].mean().reset_index()
avg_enrol = avg_enrol.rename(columns={'enrolment_2015': 'Average_Enrolment'})

#countries region
count_enrol = data.groupby('Region')['Country'].count().reset_index()
count_enrol = count_enrol.rename(columns={'Country': 'Number_of_Countries'})

#merge
regional_stats = avg_enrol.merge(count_enrol, on='Region', how='inner')

regional_stats['Average_Enrolment'] = regional_stats['Average_Enrolment'].round(0).astype(int)
regional_stats = regional_stats.sort_values('Average_Enrolment', ascending=False)

#output
print("Average secondary school enrolment by region - 2015")
print(regional_stats.to_string(index=False))

leading_region = regional_stats.iloc[0]['Region']
leading_average = regional_stats.iloc[0]['Average_Enrolment']

print(f"Leading region: {leading_region}")
print(f"Average enrolment: {leading_average:,} students")
```

```python
#question 1 p

#filter
primary_mask = df_education_clean["Education_Indicator"].str.contains("Enrolment in primary education", case=False, na=False)
primary_df = df_education_clean[primary_mask].copy()

#country year
primary_2015 = primary_df[["Country", "year_2015"]].copy()
primary_2015 = primary_2015.rename(columns={"year_2015": "primary_2015"})

#numeric
primary_2015["primary_2015"] = primary_2015["primary_2015"].astype(float)

#country,area
area_df = df_country[["country", "area"]].copy()
area_df = area_df.rename(columns={"country": "Country", "area": "Area"})
area_df["Area"] = area_df["Area"].astype(float)

#merge
merged_corr = primary_2015.merge(area_df, on="Country", how="inner")

#missing values
merged_corr = merged_corr.dropna(subset=["primary_2015", "Area"])

#correlation
corr_value = merged_corr["Area"].corr(merged_corr["primary_2015"])

print(f"Correlation in 2015: {corr_value}")
```
